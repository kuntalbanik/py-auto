```python
# Using of TypeAlias
# TypeAlias is used to create a type alias for a complex type
# 
# Eta use kore complex type ke short name diye represent kora jay

from typing import TypeAlias
student_marksheet: TypeAlias = dict[str,dict[str,int]]

def process_data_card(data: student_marksheet) -> None:
    for student, subjects in data.items():
        print(student,subjects)

records: student_marksheet = {
    "John": {"Math":90,"Physics": 85},
    "Jane": {"Math":78,"Physics": 92}
}

process_data_card(records)
```

```python
# TypeDict used to define a dictionary structure with specific keys and types.
# Normally any new key-value pair can be added to a dictionary
# But with TypeDict, we can restrict the keys to a specific set of keys and their types
# 

from typing import TypedDict

# Fixed structure for a person dictionary schema
class Person(TypedDict):
    name: str
    age: int
    email: str

person: Person = {
    "name": "Jane",
    "age": 30,
    "email": "john@example.com"
}
print(person)
```
